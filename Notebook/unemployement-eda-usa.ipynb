### Business Questions and Visualization

1. Which month of the year 2015 record the highest number of unemployment Rate?

```python
data=df[df['Year']==2015].groupby('Month')['Rate'].mean().sort_values(ascending=False)
data
```

```python
plt.figure(figsize=(10,5))
data.plot(kind='bar')
plt.title('Average unemployment rate by Month(2015)')
plt.xlabel('Month')
plt.ylabel('Average unemployment Rate')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

***January records the higehst number of unemployement rate?***

2] Top 10 States with the Highest Average Unemployment?

```python
t_states = (df.groupby('State')['Rate'].mean().sort_values(ascending=False).head(10))
round(t_states,2)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=t_states.values,
    y=t_states.index
)

plt.title("Top 10 States by Average Unemployment Rate")
plt.xlabel("Average Rate")
plt.ylabel("State")

plt.show()
```

***Arizona has the Highest Average Unemployment***

3] Find the average of the Unemployment rate in the year 2016?


```python
avg=df[df['Year']==2016]['Rate'].mean()
print(f'Average of Unemployment Rate: {round(avg,2):.2f}')
```

4] What is the monthly Average Unemployment Trend?

```python
m_order = ['January','February','March','April','May','June',
               'July','August','September','October','November','December']

df['Month'] = pd.Categorical(df['Month'],
                             categories=m_order,
                             ordered=True)
```

```python
monthly=round(df.groupby('Month')['Rate'].mean(),2)
monthly
```

```python
plt.figure(figsize=(12,5))

monthly.plot(marker='o')

plt.title("Average Monthly Unemployment Rate")
plt.xlabel("Month")
plt.ylabel("Average Rate")

plt.grid(True)

plt.show()
```

***January has the highest Average Unemployment Trend***
